# Data Cleaning

```python
import pandas as pd
import os
import warnings

warnings.filterwarnings("ignore")
```

```python
# Load dataframes

accident_df = pd.read_csv("../data/raw/accident.csv")
person_df = pd.read_csv("../data/raw/person.csv")
vehicle_df = pd.read_csv("../data/raw/vehicle.csv")
```

## Normalizing Datasets

```python
before = len(person_df)

# keep only records where airbag status is known (deployed or not deployed)
# remove: 97 (Not Applicable - Not in motor vehicle), 98 (Not Reported), 99 (Unknown)
person_clean = person_df[person_df['AIR_BAG'] < 97].copy()

print(f"Removed {before - len(person_clean)} rows")
```

```python
before = len(person_df)

# remove unknown ages
person_clean = person_clean[person_clean['AGE'] < 998]

print(f"Removed {before - len(person_clean)} rows")
```

```python
before = len(person_df)

# remove unknown injury severity statuses
person_clean = person_clean[person_clean['INJ_SEV'] != 9]

print(f"Removed {before - len(person_clean)} rows")
```

```python
# update airbag deployment to binary
# 20 = Not Deployed, 2-9 = Various deployment types
person_clean['AIR_BAG_DEPLOYED'] = person_clean['AIR_BAG'].apply(lambda x: 0 if x == 20 else 1)
```

```python
# condense rows to one per vehicle since there may be multiple people per vehicle
# aggregates data from each person in the vehicle into a single row

vehicle_person_agg = person_clean.groupby(['CASENUM', 'VEH_NO']).agg({
    'AIR_BAG_DEPLOYED': ['sum', 'count', 'mean'],  # sum=total deployed, count=total occupants
    'AGE': 'mean',  # average age of occupants
    'SEX': lambda x: x.mode()[0] if len(x.mode()) > 0 else x.iloc[0],  # most common sex
    'INJ_SEV': 'max',  # maximum injury severity in vehicle
    'DRINKING': 'max',  # any drinking in vehicle
    'ALC_STATUS': 'max',  # any positive alcohol status
    'SEAT_POS': 'min',  # minimum seat position 
    'REST_USE': lambda x: (x == 7).any().astype(int),  # 1 if any occupant not restrained (seatbelt)
    'EJECTION': 'max',  # maximum ejection severity
}).reset_index()
```

```python
# rename columns for easier use
vehicle_person_agg.columns = ['CASENUM', 'VEH_NO', 'AIRBAGS_DEPLOYED_COUNT', 'TOTAL_OCCUPANTS', 
                               'DEPLOYMENT_RATE', 'AVG_AGE', 'PREDOMINANT_SEX', 
                               'MAX_INJURY_SEV', 'ANY_DRINKING', 'ANY_ALCOHOL',
                               'FRONT_MOST_SEAT_POS', 'ANY_UNRESTRAINED', 'MAX_EJECTION']
```

```python
# define target variable as all airbags deployed = success
vehicle_person_agg['ALL_AIRBAGS_DEPLOYED'] = (
    vehicle_person_agg['AIRBAGS_DEPLOYED_COUNT'] == vehicle_person_agg['TOTAL_OCCUPANTS']
).astype(int)
```

## Merging Datasets

```python
accident_features = [
    'CASENUM',
    'REGIONNAME',
    'URBANICITYNAME',  
    'MONTH',
    'MONTHNAME',
    'DAY_WEEK',
    'DAY_WEEKNAME',
    'YEAR',
    'HOUR',
    'HOURNAME',
    'MINUTE',
    'MINUTENAME',
    'HARM_EV',
    'HARM_EVNAME',
    'MAN_COLL',
    'MAN_COLLNAME',
    'TYP_INT',
    'TYP_INTNAME',      
    'REL_ROAD',
    'REL_ROADNAME',
    'WEATHER',
    'WEATHERNAME',
    'LGT_COND',
    'LGT_CONDNAME',
    'NUM_INJ'
]

vehicle_features = [
    'CASENUM',
    'VEH_NO',
    'MAKE',
    'MAKENAME',
    'MOD_YEAR',  
    'MODEL',
    'VPICMODELNAME',
    'BODY_TYP',
    'BODY_TYPNAME',
    'VPICBODYCLASS',
    'VPICBODYCLASSNAME',
    'IMPACT1',
    'IMPACT1NAME',
    'P_CRASH1',
    'P_CRASH1NAME',
    'ACC_CONFIG',
    'ACC_CONFIGNAME',
    'VSPD_LIM',
    'TRAV_SP',
    'ROLLOVER',
    'ROLLOVERNAME',
    'NUMOCCS',
    'VSURCOND',
    'VSURCONDNAME',
    'VEH_ALCH',
    'VEH_ALCHNAME'
]

accident_features_df = accident_df[accident_features].copy()
vehicle_features_df = vehicle_df[vehicle_features].copy()
```

```python
merged_df = vehicle_features_df.merge(vehicle_person_agg, on=['CASENUM', 'VEH_NO'], how='inner').merge(accident_features_df, on='CASENUM', how='left')
print(f"After merging datasets: {len(merged_df)} rows")
```

## Filtering Columns

```python
# remove motorcyles from dataframe

clean_df  = merged_df[merged_df['VPICBODYCLASS'] != 12].copy()
print(f"Removed {len(merged_df) - len(clean_df)} rows")
```

```python
# only remove implicit missing values for useful features to minimize rows cleaned
before = len(clean_df)

# remove: 998 (Not Reported), 999 (Reported as Unknown)
clean_df = clean_df[clean_df['TRAV_SP'] < 998]
print(f"Removed {before - len(clean_df)} rows")
```

```python
before = len(clean_df)

clean_df = clean_df[clean_df['MAKENAME'] != "Not Reported"]
print(f"Removed {before - len(clean_df)} rows")
```

```python
before = len(clean_df)

clean_df = clean_df[clean_df['VPICBODYCLASS'] < 998]
print(f"Removed {before - len(clean_df)} rows")
```

## Feature Engineering

```python
# calculate vehicle age from time of accident
clean_df['VEHICLE_AGE'] = clean_df['YEAR'] - clean_df['MOD_YEAR']
```

```python
# identify if the accident was head on (frontal) by airbag deployment area
clean_df['FRONTAL_IMPACT'] = clean_df['IMPACT1'].isin(range(1, 13)).astype(int)
```

```python
# classify accidents into high speed crashes (generally accepted as over 30 mph)
clean_df['HIGH_SPEED_CRASH'] = (clean_df['TRAV_SP'] > 30).astype(int)

# modern airbags typically deploy at around 15-20 mph
# airbags deployment are based on deceleration, but generally this is the minimum range
clean_df['MIN_AIRBAG_SPEED'] = (clean_df['TRAV_SP'] > 20).astype(int)
```

```python
SEVERE_COLLISION_CODES = {
    # High-energy collisions with motor vehicles
    12,  # Collision with motor vehicle in transport
    54, 55,  # Other motor vehicle collisions
    
    # Fixed object collisions; walls, buildings, barriers
    17, 19, 20, 21, 23, 24, 25, 26,
    
    # Posts, poles, supports, trees, traffic control devices
    30, 31, 32, 33, 34, 35, 38, 39, 40, 41, 42, 43,
    
    # Other fixed objects
    46, 48, 50, 53, 57, 58, 59,
}
```

```python
#  create a composite score for the severity of a crash on a scale from 0-100
def calculate_crash_severity(row):
    score = 0
    
    # speed factor
    speed = float(row['TRAV_SP'])
    # exponential scaling for higher speeds
    if speed <= 15:
        score += speed * 0.5  # 0-7.5 points
    elif speed <= 30:
        score += 7.5 + (speed - 15) * 0.8  # 7.5-19.5 points
    elif speed <= 50:
        score += 19.5 + (speed - 30) * 1.0  # 19.5-39.5 points
    else:
        score += 39.5 + min(0.5, (speed - 50) * 0.02)  # 39.5-40 points
    
    # impact location 
    impact_code = int(row['IMPACT1'])
    
    # Categorize impact severity
    if 1 <= impact_code <= 12:  # Clock points (1-12)
        if impact_code in [12, 1, 11, 2, 10]:   # Front impacts are more severe (12, 1, 11 o'clock)
            score += 25  # Frontal/front-corner
        elif impact_code in [3, 4, 8, 9]:
            score += 20  # Side
        else:  
            score += 10  # Rear/side-rear
    elif impact_code in [13, 14]:  # Top/bottom
        score += 30  # Rollover/vertical impact 
    elif impact_code in [61, 62, 63, 81, 82, 83]:  # Left/Right (general)
        score += 15
    else: 
        score += 5 # non collision / unknown

    # type of collision
    harm_ev = int(row['HARM_EV'])
    
    if harm_ev in range(1,5):
        score += 20 # severe: rollover, fire, etc
    elif harm_ev in SEVERE_COLLISION_CODES:
        score += 15 # other vehicles or fixed objects
    else:
        score += 10 # non-fixed objects
    
    # rollovers
    if row['ROLLOVER'] == 3:
        score += 10  # Rollovers are more dangerous
    
    # scale to 0-100
    final_score = min(100, max(0, score))
    
    return final_score

clean_df['CRASH_SEVERITY_SCORE'] = clean_df.apply(calculate_crash_severity, axis=1)
```

## Final Dataset
